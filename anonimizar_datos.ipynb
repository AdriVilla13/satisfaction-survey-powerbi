```python
#Anonymization of survey data

#This script replaces server names with generic labels ("Mesero 1","Mesero 2",...) before publishing the dashboard for privacy reasons

# **Input:** encuesta_satisfaccion.xlsx , the original survey export.
# **Output:** encuesta_anonimizada.xlsx , used as the data source for the Power BI report.
```

```python
import pandas as pd
df = pd.read_excel("encuesta_satisfaccion.xlsx")
df.shape
```

```python
meseros = sorted(df["Mesero"].unique())
meseros
```

```python
mapa = {mesero: f"Mesero {i}" for i, mesero in enumerate(meseros, start=1)}
mapa
```

```python
df["Mesero"]= df["Mesero"].map(mapa)
df["Mesero"].value_counts().sort_index()
```

```python
df["Comentarios"].dropna().tolist()
```

```python
df.to_excel("encuesta_anonimizada.xlsx", sheet_name = "Respuestas de formulario", index=False)
```

```python
revision = pd.read_excel("encuesta_anonimizada.xlsx")
print(revision["Mesero"].value_counts().sort_index())
print("Vacíos:", revision["Mesero"].isna().sum())
```
